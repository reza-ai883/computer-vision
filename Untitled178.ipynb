```python
from collections import deque
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
EMOTION_LABELS = ['Angry', 'Disgust', 'Fear', 'Happy', 'Sad', 'Surprise', 'Neutral']
emotion_colors = {
    'Angry':    (0, 0, 255),     
    'Disgust':  (100, 55, 128),     
    'Fear':     (128, 0, 128),   
    'Happy':    (0, 255, 0),     
    'Sad':      (255, 0, 0),     
    'Surprise': (0, 255, 255)}    
model_path=r'C:\Users\fatima\Downloads\Facial-Emotion-Recognition-PyTorch-ONNX-master\ONNX\models\onnx_model.onnx'
```

```python
def scale_color(color, intensity):
    intensity = np.clip(intensity, 0, 1)
    return tuple(int(c * intensity + 30) for c in color)
```

```python
def draw_emotion_chart(frame, output, labels):
    chart_height, chart_width = frame.shape[0], 300
    chart_img = np.zeros((chart_height, chart_width, 3), dtype=np.uint8)
    
    bar_height = 40
    gap = 20
    
    normalized_output = output / (output.max() + 1e-6) 

    for i, label in enumerate(labels):
        p = float(output[i])
        bar_width = int(normalized_output[i] * 150)
        base_color = emotion_colors.get(label, (255, 255, 255))
        bar_color = scale_color(base_color, p)
                
        cv2.putText(chart_img, label, (10, i * (bar_height + gap) + 30), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 1)
        
        cv2.rectangle(chart_img, (100, i * (bar_height + gap) + 10), 
                      (100 + bar_width, i * (bar_height + gap) + 40), bar_color, -1)
        
    return chart_img
```

```python
def softmax(x):
    x = x - np.max(x)
    e = np.exp(x)
    return e / np.sum(e)
```

```python
detector=cv2.FaceDetectorYN_create(r'I:\v12\face_detection_yunet_2023mar.onnx',"",(320,320),0.3,0.3,5000)
recognizer=cv2.FaceRecognizerSF_create(r'I:\v12\face_recognition_sface_2021dec.onnx',"")
net=cv2.dnn.readNetFromONNX(model_path)

cv2.namedWindow('h',cv2.WINDOW_NORMAL)
cap=cv2.VideoCapture(0)
emotion_buffer = deque(maxlen=5)
while cap.isOpened:
    ret,frame=cap.read()
    if not ret:
        print('frame is not available')
        break
    frame=cv2.flip(frame,1)
    h,w=frame.shape[:2]
    
    detector.setInputSize((w,h))
    result=detector.detect(frame)
    
    if result[1] is not None:
        face=result[1][0]
        x,y,w1,h1=list(map(int,face[:4]))
        align_face=recognizer.alignCrop(frame,result[1][0])
        
        try:
            resized_frame=cv2.resize(align_face,(48,48),interpolation=cv2.INTER_AREA)
        except cv2.error as e:
            print(f'erro in load frame:{e}')
            break
            
        gray_face = cv2.cvtColor(resized_frame, cv2.COLOR_BGR2GRAY)
        blob=cv2.dnn.blobFromImage(gray_face,1.0/95,(48,48),(0))
        net.setInput(blob)
        
        try:
            output=net.forward().flatten()   
            output = softmax(output)
        except cv2.error as e:
            print(f'error in load model:{e}')
            break
            
        emotion_buffer.append(output)    
        smoothed_output = np.mean(emotion_buffer, axis=0)
        chart = draw_emotion_chart(frame, smoothed_output, EMOTION_LABELS) 
        final_display = np.hstack((frame, chart))
        
        # predict_index=np.argmax(output)
        # confidence=output[predict_index]
        # predicted_emotion=EMOTION_LABELS[predict_index]
        # confidence_percent = confidence * 100
        
        
        cv2.rectangle(final_display,(x,y),(x+w1,y+h1),(0,255,0),3)  
        final_display[y:y+h,x:x+w]=cv2.fastNlMeansDenoisingColored(final_display[y:y+h,x:x+w], None, 10, 10, 7, 21)
        cv2.imshow('h',final_display)
        cv2.resizeWindow('h',900,600) 
    else:
        print('error')
    if cv2.waitKey(5)&0xFF==27:
            break
cap.release()
cv2.destroyAllWindows()
```
